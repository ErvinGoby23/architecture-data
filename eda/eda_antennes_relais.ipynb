```python
import pandas as pd

FILE = '../../architecture-data-projet/brute/Score de connectivité/antennes-relais.csv'

df_antennes = pd.read_csv(FILE, sep=None, engine='python', encoding='utf-8')

df_antennes.columns = df_antennes.columns.str.strip().str.replace('\ufeff', '', regex=False)

print(f'Shape : {df_antennes.shape}')
print(f'Colonnes : {list(df_antennes.columns)}')
df_antennes.head()
```

```python
print(df_antennes.dtypes)
print(f'\nValeurs manquantes :')
print(df_antennes.isnull().sum())
print(f'\nArrondissements uniques : {sorted(df_antennes["Arrondissement"].unique())}')
print(f'\nOpérateurs : {df_antennes["Opérateur"].value_counts()}')
print(f'\nTypes antennes : {df_antennes["Type"].value_counts()}')
```

```python
# Codes à investiguer
print("Codes non standard (ni 75001-75020 ni 75101-75120) :")
mask = ~df_antennes['Arrondissement'].between(75001, 75020) & ~df_antennes['Arrondissement'].between(75101, 75120)
print(df_antennes[mask][['Code site', 'Adresse', 'Arrondissement', 'Opérateur']].to_string(index=False))

print(f'\nNb lignes concernées : {mask.sum()}')
```

```python
print("=== RÉSUMÉ EDA antennes-relais ===\n")
print(f"Total antennes                : {len(df_antennes)}")
print(f"Arrondissements couverts      : {df_antennes['Arrondissement'].nunique()} formats différents (75001-75020 + 75101-75120)")
print(f"1 ligne parasite              : code 76007 (Seine-Maritime) → à supprimer en silver")
print(f"\nValeurs manquantes notables :")
print(f"  Type                        : 587 NaN ({587/len(df_antennes)*100:.1f}%)")
print(f"  Mise en service             : 121 NaN")
print(f"  Mise en service 4G          : 1106 NaN")
print(f"  Mise en service 5G - 3500   : 1382 NaN")
print(f"  Mise en service 5G - 700    : 2170 NaN (100% vide → à dropper)")
print(f"  se_anno_cad_data            : 2170 NaN (100% vide → à dropper)")
print(f"\nOpérateurs :")
print(df_antennes['Opérateur'].value_counts())
print(f"\nNb antennes par arrondissement (format brut) :")
print(df_antennes.groupby('Arrondissement').size().sort_index())
```
